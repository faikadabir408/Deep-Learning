```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.datasets import mnist, fashion_mnist
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense, Dropout
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping
from sklearn.model_selection import train_test_split
import seaborn as sns


# ============================================
# 1. Autoencoder Class
# ============================================


class Autoencoder:
    def __init__(self, input_dim, encoding_dims, activation='relu',
                 output_activation='sigmoid'):
        """
        input_dim: Input dimension (e.g., 784 for 28x28 images)
        encoding_dims: List of dimensions for encoder [256, 128, 64, 32]
        """
        self.input_dim = input_dim
        self.encoding_dims = encoding_dims
        self.latent_dim = encoding_dims[-1]
        self.activation = activation
        self.output_activation = output_activation

        # Build models
        self.encoder = self._build_encoder()
        self.decoder = self._build_decoder()
        self.autoencoder = self._build_autoencoder()

        # Compile
        self.autoencoder.compile(
            optimizer=Adam(learning_rate=0.001),
            loss='mse',
            metrics=['mae']
        )

    def _build_encoder(self):
        """Build encoder network"""
        input_layer = Input(shape=(self.input_dim,))
        x = input_layer

        # Hidden layers
        for dim in self.encoding_dims:
            x = Dense(dim, activation=self.activation)(x)

        # Latent layer (bottleneck)
        latent = Dense(self.latent_dim, activation=self.activation, name='latent')(x)

        return Model(input_layer, latent, name='encoder')

    def _build_decoder(self):
        """Build decoder network"""
        latent_input = Input(shape=(self.latent_dim,))
        x = latent_input

        # Decoder layers (reverse of encoder)
        for dim in reversed(self.encoding_dims[:-1]):
            x = Dense(dim, activation=self.activation)(x)

        # Output layer
        output = Dense(self.input_dim, activation=self.output_activation)(x)

        return Model(latent_input, output, name='decoder')

    def _build_autoencoder(self):
        """Build complete autoencoder"""
        input_layer = Input(shape=(self.input_dim,))
        encoded = self.encoder(input_layer)
        decoded = self.decoder(encoded)
        return Model(input_layer, decoded, name='autoencoder')

    def train(self, X_train, X_val=None, epochs=50, batch_size=128,
              validation_split=0.2, verbose=1):
        """Train the autoencoder"""
        # Prepare data
        if X_val is None:
            history = self.autoencoder.fit(
                X_train, X_train,
                epochs=epochs,
                batch_size=batch_size,
                validation_split=validation_split,
                verbose=verbose,
                callbacks=[EarlyStopping(patience=10, restore_best_weights=True)]
            )
        else:
            history = self.autoencoder.fit(
                X_train, X_train,
                validation_data=(X_val, X_val),
                epochs=epochs,
                batch_size=batch_size,
                verbose=verbose,
                callbacks=[EarlyStopping(patience=10, restore_best_weights=True)]
            )

        return history

    def encode(self, X):
        """Encode input to latent space"""
        return self.encoder.predict(X)

    def decode(self, latent):
        """Decode from latent space"""
        return self.decoder.predict(latent)

    def reconstruct(self, X):
        """Reconstruct input"""
        return self.autoencoder.predict(X)

    def compute_compression_ratio(self):
        """Calculate compression ratio"""
        return self.input_dim / self.latent_dim

    def model_summary(self):
        """Print model summary"""
        print("\n" + "="*60)
        print("Autoencoder Architecture")
        print("="*60)
        print("\nEncoder Summary:")
        self.encoder.summary()
        print("\nDecoder Summary:")
        self.decoder.summary()
        print("\nCompression Ratio: {:.2f}x".format(self.compute_compression_ratio()))


# ============================================
# 2. Data Loading and Preprocessing
# ============================================


def load_and_preprocess_data(dataset='mnist'):
    """Load and preprocess dataset"""
    if dataset == 'mnist':
        (X_train, _), (X_test, _) = mnist.load_data()
    elif dataset == 'fashion_mnist':
        (X_train, _), (X_test, _) = fashion_mnist.load_data()
    else:
        raise ValueError("Dataset must be 'mnist' or 'fashion_mnist'")

    # Normalize and flatten
    X_train = X_train.astype('float32') / 255.0
    X_test = X_test.astype('float32') / 255.0
    X_train = X_train.reshape(X_train.shape[0], -1)
    X_test = X_test.reshape(X_test.shape[0], -1)

    return X_train, X_test


# ============================================
# 3. Training and Evaluation
# ============================================


print("="*60)
print("AUTOENCODER FOR IMAGE COMPRESSION")
print("="*60)


# Load data
print("\n1. Loading MNIST Dataset...")
X_train, X_test = load_and_preprocess_data('mnist')


# Split validation
X_train, X_val = train_test_split(X_train, test_size=0.2, random_state=42)


print(f"   Training samples: {X_train.shape[0]}")
print(f"   Validation samples: {X_val.shape[0]}")
print(f"   Test samples: {X_test.shape[0]}")
print(f"   Input dimension: {X_train.shape[1]}")


# ============================================
# 4. Build and Train Autoencoder
# ============================================


print("\n" + "="*60)
print("2. Building Autoencoder")
print("="*60)


# Define architecture
input_dim = 784  # 28x28
encoding_dims = [256, 128, 64, 32]  # Encoder dimensions


autoencoder = Autoencoder(
    input_dim=input_dim,
    encoding_dims=encoding_dims,
    activation='relu',
    output_activation='sigmoid'
)


# Model summary
autoencoder.model_summary()


# ============================================
# 5. Train the Model
# ============================================


print("\n" + "="*60)
print("3. Training Autoencoder")
print("="*60)


history = autoencoder.train(
    X_train, X_val,
    epochs=50,
    batch_size=128,
    verbose=1
)


# ============================================
# 6. Evaluate on Test Set
# ============================================


print("\n" + "="*60)
print("4. Evaluating Autoencoder")
print("="*60)


# Reconstruct test images
X_test_reconstructed = autoencoder.reconstruct(X_test)
test_loss = autoencoder.autoencoder.evaluate(X_test, X_test, verbose=0)


print(f"Test Loss (MSE): {test_loss[0]:.6f}")
print(f"Test MAE: {test_loss[1]:.6f}")


# ============================================
# 7. Visualizations
# ============================================


print("\n" + "="*60)
print("5. Visualizations")
print("="*60)


# Plot training history
plt.figure(figsize=(12, 4))


plt.subplot(1, 3, 1)
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.xlabel('Epochs')
plt.ylabel('MSE Loss')
plt.title('Training and Validation Loss')
plt.legend()
plt.grid(True)


# Show sample reconstructions
n_samples = 10
sample_indices = np.random.choice(len(X_test), n_samples, replace=False)
sample_images = X_test[sample_indices]
reconstructed = autoencoder.reconstruct(sample_images)


plt.subplot(1, 3, 2)
for i in range(n_samples):
    # Original
    plt.subplot(2, n_samples, i + 1)
    plt.imshow(sample_images[i].reshape(28, 28), cmap='gray')
    plt.axis('off')
    if i == 0:
        plt.title('Original')

    # Reconstructed
    plt.subplot(2, n_samples, n_samples + i + 1)
    plt.imshow(reconstructed[i].reshape(28, 28), cmap='gray')
    plt.axis('off')
    if i == 0:
        plt.title('Reconstructed')
plt.tight_layout()


# Compression ratio visualization
plt.subplot(1, 3, 3)
compression_ratio = autoencoder.compute_compression_ratio()
latent_size = autoencoder.latent_dim
plt.bar(['Input', 'Latent', 'Compression'],
        [input_dim, latent_size, compression_ratio*10],
        color=['blue', 'green', 'orange'])
plt.ylabel('Dimension')
plt.title(f'Compression: {compression_ratio:.1f}x')
plt.grid(True, axis='y')


plt.tight_layout()
plt.show()


# ============================================
# 8. Analyze Latent Space
# ============================================


print("\n" + "="*60)
print("6. Latent Space Analysis")
print("="*60)


# Encode some images
latent_representations = autoencoder.encode(X_test[:1000])


print(f"Latent space dimension: {latent_representations.shape[1]}")
print(f"Latent representation shape: {latent_representations.shape}")


# Visualize latent space (first 2 dimensions)
plt.figure(figsize=(12, 5))


plt.subplot(1, 2, 1)
plt.scatter(latent_representations[:, 0], latent_representations[:, 1],
            alpha=0.5, s=5)
plt.xlabel('Latent Dimension 1')
plt.ylabel('Latent Dimension 2')
plt.title('Latent Space Visualization (2D Projection)')
plt.grid(True)


# Distribution of latent values
plt.subplot(1, 2, 2)
plt.hist(latent_representations.flatten(), bins=50, alpha=0.7)
plt.xlabel('Latent Values')
plt.ylabel('Frequency')
plt.title('Distribution of Latent Space Values')
plt.grid(True)


plt.tight_layout()
plt.show()


# ============================================
# 9. Different Compression Ratios
# ============================================


print("\n" + "="*60)
print("7. Comparing Different Compression Ratios")
print("="*60)


# Try different latent dimensions
latent_dims = [8, 16, 32, 64, 128]
compression_results = []


plt.figure(figsize=(15, 8))


for idx, latent_dim in enumerate(latent_dims):
    print(f"\nTraining with Latent Dimension: {latent_dim}")

    # Build autoencoder with different latent dimension
    encoding_dims_compare = [256, 128, 64, latent_dim]

    autoencoder_compare = Autoencoder(
        input_dim=input_dim,
        encoding_dims=encoding_dims_compare,
        activation='relu',
        output_activation='sigmoid'
    )

    # Train for fewer epochs
    history_compare = autoencoder_compare.train(
        X_train, X_val,
        epochs=30,
        batch_size=128,
        verbose=0
    )

    # Evaluate
    test_loss_compare = autoencoder_compare.autoencoder.evaluate(X_test, X_test, verbose=0)
    compression_ratio = autoencoder_compare.compute_compression_ratio()

    compression_results.append({
        'latent_dim': latent_dim,
        'loss': test_loss_compare[0],
        'mae': test_loss_compare[1],
        'compression_ratio': compression_ratio
    })

    # Show reconstructions
    sample_idx = np.random.randint(0, len(X_test))
    original = X_test[sample_idx].reshape(28, 28)
    reconstructed = autoencoder_compare.reconstruct(X_test[sample_idx:sample_idx+1]).reshape(28, 28)

    plt.subplot(3, len(latent_dims), idx + 1)
    plt.imshow(original, cmap='gray')
    plt.title(f'Original\nDim: 784')
    plt.axis('off')

    plt.subplot(3, len(latent_dims), len(latent_dims) + idx + 1)
    plt.imshow(reconstructed, cmap='gray')
    plt.title(f'Latent: {latent_dim}\nRatio: {compression_ratio:.1f}x')
    plt.axis('off')

    plt.subplot(3, len(latent_dims), 2*len(latent_dims) + idx + 1)
    plt.bar(['Loss', 'MAE'], [test_loss_compare[0], test_loss_compare[1]])
    plt.title(f'MSE: {test_loss_compare[0]:.4f}')
    plt.grid(True, axis='y')


plt.tight_layout()
plt.show()


# ============================================
# 10. Results Summary
# ============================================


print("\n" + "="*60)
print("8. Results Summary")
print("="*60)


print("\nCompression Results Comparison:")
print("-" * 60)
print(f"{'Latent Dim':<12} {'Compression Ratio':<20} {'MSE Loss':<12} {'MAE':<12}")
print("-" * 60)


for result in compression_results:
    print(f"{result['latent_dim']:<12} {result['compression_ratio']:.1f}x{' ':<17} "
          f"{result['loss']:.6f}  {result['mae']:.6f}")


print("\n" + "="*60)
print("9. Model Architecture Details")
print("="*60)


print(f"\nInput Dimension: {input_dim}")
print(f"Encoder Layers: {encoding_dims}")
print(f"Latent Dimension: {autoencoder.latent_dim}")
print(f"Total Parameters:")
encoder_params = autoencoder.encoder.count_params()
decoder_params = autoencoder.decoder.count_params()
total_params = encoder_params + decoder_params
print(f"  Encoder: {encoder_params:,}")
print(f"  Decoder: {decoder_params:,}")
print(f"  Total: {total_params:,}")


# ============================================
# 11. Fashion MNIST Example
# ============================================


print("\n" + "="*60)
print("10. Fashion MNIST Autoencoder")
print("="*60)


# Load Fashion MNIST
X_train_fashion, X_test_fashion = load_and_preprocess_data('fashion_mnist')
X_train_fashion, X_val_fashion = train_test_split(X_train_fashion, test_size=0.2, random_state=42)


# Build autoencoder for Fashion MNIST
autoencoder_fashion = Autoencoder(
    input_dim=784,
    encoding_dims=[256, 128, 64, 32],
    activation='relu',
    output_activation='sigmoid'
)


# Train
history_fashion = autoencoder_fashion.train(
    X_train_fashion, X_val_fashion,
    epochs=30,
    batch_size=128,
    verbose=0
)


# Reconstruct
X_fashion_reconstructed = autoencoder_fashion.reconstruct(X_test_fashion[:10])


# Visualize Fashion MNIST results
plt.figure(figsize=(15, 4))


for i in range(10):
    # Original
    plt.subplot(2, 10, i + 1)
    plt.imshow(X_test_fashion[i].reshape(28, 28), cmap='gray')
    plt.axis('off')
    if i == 0:
        plt.title('Original', fontsize=12)

    # Reconstructed
    plt.subplot(2, 10, 10 + i + 1)
    plt.imshow(X_fashion_reconstructed[i].reshape(28, 28), cmap='gray')
    plt.axis('off')
    if i == 0:
        plt.title('Reconstructed', fontsize=12)


plt.suptitle('Fashion MNIST Autoencoder Results', fontsize=14)
plt.tight_layout()
plt.show()


print(f"Fashion MNIST Test MSE: {autoencoder_fashion.autoencoder.evaluate(X_test_fashion, X_test_fashion, verbose=0)[0]:.6f}")
print(f"Compression Ratio: {autoencoder_fashion.compute_compression_ratio():.1f}x")
```
